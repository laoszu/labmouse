```python
import json
import matplotlib.pyplot as plt
from matplotlib.patches import Rectangle
from PIL import Image
import numpy as np
import pandas as pd
import random
```

```python
def display_sample(json_path, data_path, limit):
    with open(json_path) as json_file:
        data = json.load(json_file)

    filenames = list(data.keys())
    n_image = random.randint(0, limit)
    filename = filenames[n_image]
    box = data[filename][0]['Boundingbox']
    print(data[filename])

    box = {key: int(value) for key, value in box.items()}
    print(box)

    img = np.asarray(Image.open(f"{data_path}/{filename}"))
    plt.title(filename)
    plt.imshow(img)

    ax = plt.gca()
    rect = Rectangle((box['x'], box['y']), box['w'] - box['x'], box['h'] - box['y'], linewidth=1, edgecolor='r', facecolor='none')
    ax.add_patch(rect)
```

```python
display_sample("data/face_detection_data.json", "Face_detection", 2000)
```

```python
display_sample("data/eyes_detection_data.json", "Eyes_detection", 1000)
```
